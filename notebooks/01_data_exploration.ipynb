```python
import geopandas as gpd

gdf = gpd.read_file("../data/geo/BR_Municipios_2022.shp")

gdf.head()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(15, 15))

gdf.plot(ax=ax)

plt.title("Municípios do Brasil - IBGE 2022")
plt.show()
```

```python
print(f"Municípios: {len(gdf)}")

print("\nEstados:")
print(gdf["SIGLA_UF"].unique())

print("\nColunas:")
print(gdf.columns)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 12))

gdf.plot(
    ax=ax,
    linewidth=0.1,
    edgecolor="black"
)

plt.title("Brasil - Municípios IBGE 2022")
plt.axis("off")

plt.show()
```
